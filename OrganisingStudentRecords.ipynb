***Organizing Student Records***

**Objective**

To improve the project by organizing each student's academic information into a single record and processing these records using a loop.

**Theory**

As the number of students increases, keeping separate lists for names and each subject becomes difficult to manage. The project needs a better way to keep the information belonging to one student together.

A tuple can store related values as one collection. In this stage, each student's name and subject marks will be grouped into a tuple. The project will then use a for loop to access each student's record, calculate the total and average marks, and determine the performance category.

This makes the student data more organized and prepares the project for handling larger datasets later.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    ("Rahul", 78, 82, 75),
    ("Priya", 88, 91, 85),
    ("Aman", 65, 70, 68)
]

# Process each student record
for student in student_records:

    name = student[0]
    python_marks = student[1]
    maths_marks = student[2]
    dbms_marks = student[3]

    # Calculate total and average
    total_marks = python_marks + maths_marks + dbms_marks
    average_marks = total_marks / 3

    # Evaluate performance
    if average_marks >= 80:
        performance = "Excellent"
    elif average_marks >= 60:
        performance = "Good"
    elif average_marks >= 40:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    print("\nStudent:", name)
    print("Python Marks:", python_marks)
    print("Maths Marks:", maths_marks)
    print("DBMS Marks:", dbms_marks)
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Performance:", performance)
```
