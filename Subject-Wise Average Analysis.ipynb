***Subject-Wise Average Analysis***

**Objective**

To extend the project by calculating the average marks in each subject and identifying which subject has the highest class average.

**Theory**

The project currently analyzes students individually. However, teachers may also want to know how the entire class performs in each subject.
In this stage, the project calculates the class average for Python, Maths, and DBMS. It adds the marks of each subject across all students and divides the total by the number of students.
The results help identify the strongest-performing subject in the class and subjects where students may need more practice.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85},
    {"Name": "Aman", "Python": 35, "Maths": 42, "DBMS": 38},
    {"Name": "Sneha", "Python": 55, "Maths": 58, "DBMS": 52}
]

# Subject names
subjects = ["Python", "Maths", "DBMS"]

# Calculate subject-wise averages
subject_averages = {}

for subject in subjects:
    total = 0

    for student in student_records:
        total += student[subject]

    average = total / len(student_records)
    subject_averages[subject] = average

# Display subject-wise averages
print("\nSubject-Wise Class Average")
print("--------------------------")

for subject, average in subject_averages.items():
    print(subject + ":", round(average, 2))

# Identify the highest class average
best_subject = max(subject_averages, key=subject_averages.get)

print("\nSubject with Highest Class Average")
print("----------------------------------")
print("Subject:", best_subject)
print("Average Marks:", round(subject_averages[best_subject], 2))
```

- Dictionary items() method
- max() function
