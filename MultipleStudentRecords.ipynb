***Multiple Student Records***
**Objective**

To extend the project by using a loop to process the marks of multiple students and calculate their average performance.

**Theory**

So far, the project has analyzed only one student. To make the analyzer more useful, it should be able to process information for multiple students without writing the same calculation code repeatedly.

A for loop allows the project to repeat the same calculation for each student's marks. In this stage, the project will process the marks of three students, calculate their total and average marks, and display the results.

This is the first step toward turning the project from a single-student calculator into a student performance analyzer.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student names
students = ["Rahul", "Priya", "Aman"]

# Marks of each student
python_marks = [78, 88, 65]
maths_marks = [82, 91, 70]
dbms_marks = [75, 85, 68]

# Process each student
for i in range(3):

    total_marks = python_marks[i] + maths_marks[i] + dbms_marks[i]
    average_marks = total_marks / 3

    # Evaluate performance
    if average_marks >= 80:
        performance = "Excellent"
    elif average_marks >= 60:
        performance = "Good"
    elif average_marks >= 40:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    print("\nStudent:", students[i])
    print("Python Marks:", python_marks[i])
    print("Maths Marks:", maths_marks[i])
    print("DBMS Marks:", dbms_marks[i])
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Performance:", performance)
```
