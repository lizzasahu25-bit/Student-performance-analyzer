***Handling Missing Data Using Pandas***

**1. Objective**

To identify and handle missing values in the student dataset using Pandas so that incomplete records do not affect data analysis.

**2. Theory**

Sometimes student records may contain missing information, such as marks or attendance that were not entered correctly.

Missing data can affect calculations and produce inaccurate results.

In this project, we will use Pandas to:
- Check whether any values are missing.
- Count missing values in each column.
- Replace missing numerical values with the average of that column.
- Verify that the dataset no longer contains missing values.
We will use isnull(), sum(), and fillna() methods to handle missing information.

**3. Code**

```python
import pandas as pd

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Create a mock dataset file for demonstration
mock_data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Python": [85.0, 90.0, 78.0, 92.0, 88.0],
    "Attendance": [95.0, 80.0, 85.0, 90.0, 95.0]
}
df_initial = pd.DataFrame(mock_data)
df_initial.to_csv("student_records.csv", index=False)

# Load student records
df = pd.read_csv("student_records.csv")

# Introduce missing values for testing
df.loc[1, "Python"] = None
df.loc[2, "Attendance"] = None

# Display dataset with missing values
print("\nDataset Before Cleaning:")
print(df)

# Check missing values
print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# Fill missing numerical values with column averages
df["Python"] = df["Python"].fillna(df["Python"].mean())

df["Attendance"] = df["Attendance"].fillna(
    df["Attendance"].mean()
)

# Display cleaned dataset
print("\nDataset After Cleaning:")
print(df)

# Verify missing values
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# Save cleaned dataset
df.to_csv("cleaned_student_records.csv", index=False)

print("\nCleaned student data saved successfully!")
```

**Concepts Used**
1. isnull(): Identifies missing values in a dataset.
2. sum(): Counts missing values in each column.
3. fillna(): Replaces missing values with specified values.
4. mean(): Calculates the average of a numerical column.
5. to_csv(): Saves the cleaned dataset into a new CSV file.
