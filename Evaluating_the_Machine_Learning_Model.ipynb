Evaluating the Machine Learning Model
1. Objective
To evaluate the trained Decision Tree model and check how accurately it predicts student performance using testing data.
2. Theory
After training a Machine Learning model, we need to check whether it can make correct predictions.

This process is called Model Evaluation.

In this project, we will use the trained Decision Tree model to predict the performance categories of students in the testing dataset.

We will compare the predicted results with the actual performance categories to calculate the model's accuracy.

We will use the accuracy_score() function from Scikit-learn.

Accuracy tells us how many predictions made by the model are correct compared to the total number of predictions.
3. Code

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Load student records
df = pd.read_csv("student_records.csv")

# Add study hours
df["Study_Hours"] = [3, 5, 1, 2]

# Calculate average marks
df["Average"] = (
    df["Python"] + df["Maths"] + df["DBMS"]
) / 3

# Create performance categories
def classify_performance(average):

    if average >= 80:
        return "Excellent"
    elif average >= 60:
        return "Good"
    elif average >= 40:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average"].apply(
    classify_performance
)

# Select input and target
X = df[["Study_Hours", "Attendance"]]
y = df["Performance"]

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Create and train model
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# Predict testing data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Display actual and predicted results
print("\nActual Performance:")
print(list(y_test))

print("\nPredicted Performance:")
print(list(y_pred))

print("\nModel Accuracy:", accuracy * 100, "%")
```

**Concepts Used**
1. Model Evaluation: Checks how well a trained model performs.
2. predict(): Generates predictions using a trained model.
3. accuracy_score(): Calculates the percentage of correct predictions.
4. Actual Values: The original performance categories.
5. Predicted Values: The categories estimated by the model.
