***Input Validation***

**1. Objective**

To validate student details before adding them to the project database and prevent incorrect data from being entered.

**2. Theory**

While adding student records, users may enter incorrect information, such as negative marks, attendance above 100%, or an empty name.

Input validation helps ensure that only valid information is stored.

In this project, we will apply three validation rules:
- Student name must not be empty.
- Subject marks must be between 0 and 100.
- Attendance must be between 0 and 100.
A while loop will repeatedly ask for input until the user enters a valid value.

Code:**

````python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Existing student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82,
     "DBMS": 75, "Attendance": 85},

    {"Name": "Priya", "Python": 88, "Maths": 91,
     "DBMS": 85, "Attendance": 92},

    {"Name": "Aman", "Python": 35, "Maths": 42,
     "DBMS": 38, "Attendance": 68},

    {"Name": "Sneha", "Python": 55, "Maths": 58,
     "DBMS": 52, "Attendance": 72}
]

# Validate student name
while True:
    name = input("Enter student name: ").strip()

    if name != "":
        break
    else:
        print("Name cannot be empty. Try again.")

# Function to validate marks and attendance
def get_valid_input(message):
    while True:
        value = int(input(message))

        if 0 <= value <= 100:
            return value
        else:
            print("Enter a value between 0 and 100.")

# Get valid student details
python_marks = get_valid_input("Enter Python marks: ")
maths_marks = get_valid_input("Enter Maths marks: ")
dbms_marks = get_valid_input("Enter DBMS marks: ")
attendance = get_valid_input("Enter attendance percentage: ")

# Create student record
new_student = {
    "Name": name,
    "Python": python_marks,
    "Maths": maths_marks,
    "DBMS": dbms_marks,
    "Attendance": attendance
}

# Add record to database
student_records.append(new_student)

print("\nStudent added successfully!")
print("Name:", name)
print("Python Marks:", python_marks)
print("Maths Marks:", maths_marks)
print("DBMS Marks:", dbms_marks)
print("Attendance:", attendance, "%")

print("\nTotal Student Records:", len(student_records))
````

**Concepts Used**
1. While Loop: Repeats input until a valid value is entered.
2. Conditional Statements: Check whether the entered values are valid.
3. Functions: get_valid_input() performs input validation.
4. Return Statement: Sends the validated value back to the main program.
5. Input Validation: Prevents incorrect student information from being stored.
