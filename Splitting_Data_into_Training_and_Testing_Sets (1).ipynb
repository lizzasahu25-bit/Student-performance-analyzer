***Splitting Data into Training and Testing Sets***

**1. Objective**

To divide the student dataset into training and testing sets so that it can be used to build and evaluate a Machine Learning model.

**2. Theory**

Before building a Machine Learning model, we need to divide our dataset into two parts:
- Training Data: Used to teach the model by providing examples of student information and their performance categories.
- Testing Data: Used to check how accurately the model predicts the performance of students it has not trained on.
In this project, we will use the train_test_split() function from Scikit-learn.

We will divide the dataset into 80% training data and 20% testing data.

This helps us evaluate whether the model can make predictions using new student information.

**3. Code**

```python
import pandas as pd
import os
from sklearn.model_selection import train_test_split

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Load student records, or create them if the file is missing
csv_path = "student_records.csv"
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
else:
    print(f"Warning: '{csv_path}' not found. Creating mock student dataset...")
    data = {
        "Python": [85, 70, 50, 90],
        "Maths": [90, 60, 45, 95],
        "DBMS": [80, 65, 40, 85],
        "Attendance": [92, 85, 75, 98]
    }
    df = pd.DataFrame(data)
    df.to_csv(csv_path, index=False)

# Add study hours
df["Study_Hours"] = [3, 5, 1, 2]

# Calculate average marks
df["Average"] = (
    df["Python"] + df["Maths"] + df["DBMS"]
) / 3

# Create performance categories
def classify_performance(average):
    if average >= 80:
        return "Excellent"
    elif average >= 60:
        return "Good"
    elif average >= 40:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average"].apply(
    classify_performance
)

# Select input features and target
X = df[["Study_Hours", "Attendance"]]
y = df["Performance"]

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

# Display dataset sizes
print("\nTraining Data:")
print(X_train)

print("\nTesting Data:")
print(X_test)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))
```

**Concepts Used**
1. train_test_split(): Divides a dataset into training and testing sets.
2. Training Data: Data used to train the Machine Learning model.
3. Testing Data: Data used to evaluate model performance.
4. test_size: Specifies the proportion of data reserved for testing.
5. random_state: Keeps the random split consistent when the code is run again.
