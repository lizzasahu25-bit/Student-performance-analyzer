***Saving Student Data to a CSV File***

**1. Objective**

To store student records permanently in a CSV file so that the data can be accessed even after the program stops running.

**2. Theory**

Until now, student records have been stored in a Python list. However, this data disappears when the program stops.

To solve this problem, we will use a CSV (Comma-Separated Values) file to store student information permanently.

In this project:
- We will use Python's built-in csv module.
- Each student's details will be stored as a row.
- Column names will represent student information.
- The CSV file will be created automatically if it does not exist.
This will help us maintain a permanent student database.

**3. Code**

```python
import csv

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82,
     "DBMS": 75, "Attendance": 85},

    {"Name": "Priya", "Python": 88, "Maths": 91,
     "DBMS": 85, "Attendance": 92},

    {"Name": "Aman", "Python": 35, "Maths": 42,
     "DBMS": 38, "Attendance": 68},

    {"Name": "Sneha", "Python": 55, "Maths": 58,
     "DBMS": 52, "Attendance": 72}
]

# CSV column names
field_names = [
    "Name", "Python", "Maths", "DBMS", "Attendance"
]

# Save records into CSV file
with open("student_records.csv", "w", newline="") as file:

    writer = csv.DictWriter(file, fieldnames=field_names)

    # Write column headings
    writer.writeheader()

    # Write student records
    writer.writerows(student_records)

print("\nStudent records saved successfully!")

print("File Name: student_records.csv")
print("Total Records Saved:", len(student_records))
```

**Concepts Used**

1. CSV Module: Helps read and write CSV files.
2. DictWriter(): Writes dictionary records into a CSV file.
3. Writeheader(): Adds column names to the file.
4. Writerows(): Writes multiple student records at once.
5. With Open(): Opens and manages the file safely.
