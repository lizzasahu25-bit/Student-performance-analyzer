***Identifying Students Who Need Support***

**Objective**

To extend the project by identifying students whose average marks are below 60 and displaying a list of students who may need additional academic support.

**Theory**

The project can already calculate student averages, classify performance, and identify the class topper. The next step is to find students who may need extra help.
In this stage, the project uses a list to store the names of students whose average marks are below 60. During the loop, each student's average is checked using a conditional statement. If the average is below 60, the student's name is added to the list using append().
This feature helps the analyzer identify students who may benefit from additional study or guidance.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85},
    {"Name": "Aman", "Python": 35, "Maths": 42, "DBMS": 38},
    {"Name": "Sneha", "Python": 55, "Maths": 58, "DBMS": 52}
]

# Store students who need support
students_needing_support = []

print("\nStudent Performance")
print("-------------------")

# Analyze each student
for student in student_records:

    name = student["Name"]

    total_marks = (
        student["Python"]
        + student["Maths"]
        + student["DBMS"]
    )

    average_marks = total_marks / 3

    if average_marks >= 80:
        performance = "Excellent"
    elif average_marks >= 60:
        performance = "Good"
    elif average_marks >= 40:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    print("\nStudent:", name)
    print("Total Marks:", total_marks)
    print("Average Marks:", round(average_marks, 2))
    print("Performance:", performance)

    # Identify students needing support
    if average_marks < 60:
        students_needing_support.append(name)

# Display students needing support
print("\nStudents Needing Academic Support")
print("---------------------------------")

if students_needing_support:
    for name in students_needing_support:
        print(name)
else:
    print("No students currently need additional support.")
```

**Concept Used**
- Lists
- append() method
- for loop
- Conditional statements
- Comparison operator (<)
- if and else
