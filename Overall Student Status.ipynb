***Overall Student Status***

**Objective**

To combine academic results and attendance into one overall status, helping identify students who are performing well and those who may need attention.

**Theory**

The project can now evaluate students' marks, pass/fail results, and attendance separately. In this stage, these results are combined to generate an overall status for each student.
The project uses conditional statements and the logical and operator. A student is marked as “Good Standing” if they pass all subjects, achieve an average of at least 60, and have attendance of at least 75%. Otherwise, the project marks them as “Needs Attention.”
This gives the analyzer a more complete view of each student's academic progress.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {
        "Name": "Rahul",
        "Python": 78,
        "Maths": 82,
        "DBMS": 75,
        "Attendance": 85
    },
    {
        "Name": "Priya",
        "Python": 88,
        "Maths": 91,
        "DBMS": 85,
        "Attendance": 92
    },
    {
        "Name": "Aman",
        "Python": 35,
        "Maths": 42,
        "DBMS": 38,
        "Attendance": 68
    },
    {
        "Name": "Sneha",
        "Python": 55,
        "Maths": 58,
        "DBMS": 52,
        "Attendance": 72
    }
]

subjects = ["Python", "Maths", "DBMS"]
passing_marks = 40
required_attendance = 75
required_average = 60

print("\nOverall Student Status")
print("----------------------")

for student in student_records:

    name = student["Name"]
    total_marks = 0
    passed_all_subjects = True

    # Calculate marks and check each subject
    for subject in subjects:
        marks = student[subject]
        total_marks += marks

        if marks < passing_marks:
            passed_all_subjects = False

    average_marks = total_marks / len(subjects)
    attendance = student["Attendance"]

    # Evaluate overall status
    if (
        passed_all_subjects
        and average_marks >= required_average
        and attendance >= required_attendance
    ):
        overall_status = "Good Standing"
    else:
        overall_status = "Needs Attention"

    print("\nStudent:", name)
    print("Average Marks:", round(average_marks, 2))
    print("Attendance:", attendance, "%")
    print("Overall Status:", overall_status)
```

Concept Used
- Conditional statements (if, else)
- Logical operator (and)
- Comparison operators (>=, <)
- Dictionaries and lists
- Nested for loops
- Combining multiple conditions
