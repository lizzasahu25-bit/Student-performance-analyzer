Objective

To extend the project by using operators to calculate the student's total marks and average marks from the academic data already stored.

Theory

The project now has individual student marks as input. The next step is to perform calculations on this data.

Arithmetic operators are used to perform mathematical calculations. In this stage, the project uses addition (+) to calculate total marks and division (/) to calculate the average marks.

The average marks provide a basic measure of the student's overall academic performance and will later be useful when the project evaluates and predicts student performance.

Code:

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student information
student_name = "Rahul"
attendance = 85
study_hours = 4.5

# Subject marks
python_marks = 78
maths_marks = 82
dbms_marks = 75

# Calculate total and average
total_marks = python_marks + maths_marks + dbms_marks
average_marks = total_marks / 3

print("\nStudent Information")
print("-------------------")
print("Name:", student_name)
print("Attendance:", attendance, "%")
print("Study Hours:", study_hours)

print("\nAcademic Performance")
print("--------------------")
print("Python Marks:", python_marks)
print("Maths Marks:", maths_marks)
print("DBMS Marks:", dbms_marks)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
```

Concept Used
Addition (+) → calculates total marks.
Division (/) → calculates average marks.
Variables → store subject marks and calculated results.
Arithmetic expression → combines values to produce a result.
