***Visualizing Student Performance Distribution***

**1. Objective**

To visualize the distribution of students according to their average marks using a histogram.

**2. Theory**

In the previous days, we created bar charts to compare student marks and subject-wise averages.

Now, we will use a Histogram to understand how student marks are distributed across different score ranges.

A histogram groups numerical values into intervals called bins. It helps us understand how many students fall within each marks range.

In this project, we will:
- Calculate the average marks of each student.
- Divide the averages into different score ranges.
- Create a histogram to display the distribution.
- Understand how many students belong to each performance range.

**3. Code**

```python
import pandas as pd
import matplotlib.pyplot as plt

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Load student records
df = pd.read_csv("student_records.csv")

# Calculate average marks of each student
df["Average"] = (
    df["Python"] + df["Maths"] + df["DBMS"]
) / 3

# Display student averages
print("\nStudent Average Marks:")

print(df[["Name", "Average"]])

# Create histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df["Average"],
    bins=[0, 40, 60, 70, 80, 100],
    edgecolor="black"
)

# Add chart details
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Performance")

# Display chart
plt.tight_layout()
plt.show()
```

The histogram will show:

- 0–40 marks: 1 student
- 40–60 marks: 1 student
- 60–70 marks: 0 students
- 70–80 marks: 1 student
- 80–100 marks: 1 student

**5. Concepts Used**
1. Histogram: Represents the distribution of numerical data.
2. hist(): Creates a histogram.
3. Bins: Divide numerical values into different intervals.
4. DataFrame Column: Stores the calculated average marks.
5. xlabel() and ylabel(): Add labels to the graph axes.
6. show(): Displays the generated chart.
