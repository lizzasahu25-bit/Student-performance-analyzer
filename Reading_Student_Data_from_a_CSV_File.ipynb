***Reading Student Data from a CSV File***

**1. Objective**

To read previously saved student records from a CSV file and display them in the project.

**2. Theory**

In Day 27, we created a CSV file to store student records permanently. Now, we will learn how to retrieve that stored information.

Reading data from a CSV file allows the project to access existing student records without entering them again.

In this project, we will:
- Open the existing CSV file.
- Read student information using Python.
- Store the retrieved records in a list.
- Display each student's details.
- Calculate the total number of records.
We will use csv.DictReader() to read the file because it allows us to access student details using column names.

**3. Code**

```python
import csv

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# List to store student records
student_records = []

# Read student data from CSV file
with open("student_records.csv", "r") as file:

    reader = csv.DictReader(file)

    for row in reader:
        student_records.append(row)

# Display student records
print("\nStudent Records")
print("----------------")

for student in student_records:

    print("\nName:", student["Name"])
    print("Python Marks:", student["Python"])
    print("Maths Marks:", student["Maths"])
    print("DBMS Marks:", student["DBMS"])
    print("Attendance:", student["Attendance"], "%")

# Display total records
print("\nTotal Students:", len(student_records))
```

Concepts Used
1. CSV Module: Helps retrieve information from CSV files.
2. DictReader(): Reads CSV data as dictionaries.
3. File Reading Mode (r): Opens a file for reading.
4. Append Method: Adds retrieved records to the student list.
5. For Loop: Displays each student's information.
