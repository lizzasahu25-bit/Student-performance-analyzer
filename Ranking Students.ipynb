***Ranking Students by Average Marks***

**Objective**

To extend the project by ranking students according to their average marks, from the highest-performing student to the lowest-performing student.

**Theory**

The project can now calculate individual averages and overall class statistics. The next step is to arrange students in order of their academic performance.
In this stage, the project uses Python's sorted() function to arrange student records based on their average marks in descending order. The reverse=True argument places the highest average first.
This feature generates a class ranking that helps identify the top-performing students and compare academic results.

**Code**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75,
     "Attendance": 85},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85,
     "Attendance": 92},
    {"Name": "Aman", "Python": 35, "Maths": 42, "DBMS": 38,
     "Attendance": 68},
    {"Name": "Sneha", "Python": 55, "Maths": 58, "DBMS": 52,
     "Attendance": 72}
]

subjects = ["Python", "Maths", "DBMS"]

# Calculate each student's average
for student in student_records:
    total_marks = 0

    for subject in subjects:
        total_marks += student[subject]

    student["Average"] = total_marks / len(subjects)

# Sort students by average marks
ranked_students = sorted(
    student_records,
    key=lambda student: student["Average"],
    reverse=True
)

# Display student rankings
print("\nStudent Ranking")
print("---------------")

for rank, student in enumerate(ranked_students, start=1):
    print("\nRank:", rank)
    print("Name:", student["Name"])
    print("Average Marks:", round(student["Average"], 2))
```

**Concept Used**
- sorted() function
- lambda function
- reverse=True
- enumerate() function
- Dictionaries and lists
- Nested for loops
- Descending order sorting
