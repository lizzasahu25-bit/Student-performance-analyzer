***Calculating Student Pass/Fail Status***

**Objective**

To extend the project by checking whether each student has passed or failed based on their marks in all three subjects.

**Theory**

The project can already calculate total marks, average marks, and performance categories. However, a student may have a reasonable average but still score poorly in one subject.
In this stage, the project checks each subject's marks against a minimum passing mark of 40. If the student scores at least 40 in every subject, the student is marked as “Pass.” If any subject score is below 40, the result is “Fail.”
This adds a subject-wise pass/fail check to the existing performance analysis.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85},
    {"Name": "Aman", "Python": 35, "Maths": 42, "DBMS": 38},
    {"Name": "Sneha", "Python": 55, "Maths": 58, "DBMS": 52}
]

subjects = ["Python", "Maths", "DBMS"]
passing_marks = 40

print("\nStudent Pass/Fail Report")
print("------------------------")

for student in student_records:

    name = student["Name"]

    total_marks = 0
    passed_all_subjects = True

    # Check each subject
    for subject in subjects:
        marks = student[subject]
        total_marks += marks

        if marks < passing_marks:
            passed_all_subjects = False

    average_marks = total_marks / len(subjects)

    if passed_all_subjects:
        result = "Pass"
    else:
        result = "Fail"

    print("\nStudent:", name)
    print("Total Marks:", total_marks)
    print("Average Marks:", round(average_marks, 2))
    print("Result:", result)

    # Show subjects with marks below the passing mark
    if not passed_all_subjects:
        print("Subjects Requiring Improvement:")

        for subject in subjects:
            if student[subject] < passing_marks:
                print(subject, "-", student[subject], "marks")
```

Concept Used
- Nested for loops
- Dictionaries and lists
- Conditional statements
- Boolean values (True and False)
- Logical negation (not)
- Pass/fail decision-making
