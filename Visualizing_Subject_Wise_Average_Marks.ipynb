***Visualizing Subject-Wise Average Marks***

**1. Objective**

To calculate and visualize the average marks obtained by students in each subject using a bar chart.

**2. Theory**

In Day 32, we created a grouped bar chart to compare individual students' marks across different subjects.

Now, we will focus on the overall performance of the class in each subject.

Subject-wise average analysis helps us identify which subject has the highest average marks and which subject requires more attention.

In this project, we will:
- Load student records using Pandas.
- Calculate the average marks for each subject.
- Create a bar chart to represent the averages.
- Highlight the subject with the highest average.
We will use Matplotlib to present the calculated results visually.

**3. Code**

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Create a mock dataset if 'student_records.csv' does not exist
if not os.path.exists("student_records.csv"):
    mock_data = {
        "Student": ["Alice", "Bob", "Charlie", "David", "Emma"],
        "Python": [85, 90, 78, 92, 88],
        "Maths": [75, 80, 85, 70, 95],
        "DBMS": [80, 85, 90, 88, 84]
    }
    pd.DataFrame(mock_data).to_csv("student_records.csv", index=False)
    print("Note: 'student_records.csv' was not found. A mock dataset has been created for demonstration.\n")

# Load student records
df = pd.read_csv("student_records.csv")

# Calculate subject-wise averages
subjects = ["Python", "Maths", "DBMS"]

average_marks = [
    df["Python"].mean(),
    df["Maths"].mean(),
    df["DBMS"].mean()
]

# Find the subject with the highest average
best_subject = subjects[average_marks.index(max(average_marks))]

print("Subject-Wise Average Marks:")

for i in range(len(subjects)):
    print(subjects[i], ":", round(average_marks[i], 2))

print("\nBest Performing Subject:", best_subject)

# Create bar chart
plt.figure(figsize=(8, 5))

plt.bar(subjects, average_marks, color=['#4CAF50', '#2196F3', '#FF9800'])

# Add chart details
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks in Each Subject")
plt.ylim(0, 100)

# Display chart
plt.tight_layout()
plt.show()
```

**Concepts Used**
1. mean(): Calculates the average marks in each subject.
2. List: Stores subject names and their average marks.
3. index(): Finds the position of the highest average.
4. bar(): Creates a bar chart.
5. xlabel() and ylabel(): Add labels to the graph axes.
6. title(): Adds a title to the chart.
