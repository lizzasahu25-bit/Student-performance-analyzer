***Calculating Attendance Status***

**Objective**

To extend the project by storing student attendance and identifying students whose attendance is below the required 75%.

**Theory**

Academic marks alone do not provide a complete picture of student performance. Attendance is another useful factor to consider when analyzing students' academic progress.

In this stage, the project adds an attendance value to each student's dictionary record. The program checks whether each student's attendance is at least 75%. Students meeting this requirement are marked as having “Satisfactory” attendance, while those below it are marked as “Low.”
The analyzer will now combine attendance evaluation with the existing academic performance and pass/fail analysis.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records with attendance
student_records = [
    {
        "Name": "Rahul",
        "Python": 78,
        "Maths": 82,
        "DBMS": 75,
        "Attendance": 85
    },
    {
        "Name": "Priya",
        "Python": 88,
        "Maths": 91,
        "DBMS": 85,
        "Attendance": 92
    },
    {
        "Name": "Aman",
        "Python": 35,
        "Maths": 42,
        "DBMS": 38,
        "Attendance": 68
    },
    {
        "Name": "Sneha",
        "Python": 55,
        "Maths": 58,
        "DBMS": 52,
        "Attendance": 72
    }
]

subjects = ["Python", "Maths", "DBMS"]
passing_marks = 40
required_attendance = 75

print("\nStudent Academic and Attendance Report")
print("---------------------------------------")

for student in student_records:

    name = student["Name"]
    attendance = student["Attendance"]

    # Calculate total and average marks
    total_marks = 0
    passed_all_subjects = True

    for subject in subjects:
        marks = student[subject]
        total_marks += marks

        if marks < passing_marks:
            passed_all_subjects = False

    average_marks = total_marks / len(subjects)

    # Check pass/fail status
    if passed_all_subjects:
        result = "Pass"
    else:
        result = "Fail"

    # Check attendance
    if attendance >= required_attendance:
        attendance_status = "Satisfactory"
    else:
        attendance_status = "Low"

    print("\nStudent:", name)
    print("Average Marks:", round(average_marks, 2))
    print("Academic Result:", result)
    print("Attendance:", attendance, "%")
    print("Attendance Status:", attendance_status)
```

**Concept Used**
- Dictionaries
- Lists
- Nested for loops
- Conditional statements
- Comparison operators (>=, <)
- Boolean values
- Attendance evaluation
