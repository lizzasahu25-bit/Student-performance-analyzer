**Training a Machine Learning Model**

**1. Objective**

To train a Machine Learning model using student records and teach it to classify students into different academic performance categories.

**2. Theory**

In the previous days, we prepared the student dataset and divided it into training and testing sets.

Now, we will train our first Machine Learning model.

We will use the Decision Tree Classifier, a supervised Machine Learning algorithm that makes predictions by learning patterns from existing data.

In our project:
- Study hours and attendance will be used as input features.
- Performance categories will be used as target outputs.
- The model will learn the relationship between these inputs and outputs.
- After training, the model will be ready to predict student performance.
A Decision Tree works like a flowchart, making decisions based on conditions in the data.

**3. Code**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Create a dummy student_records.csv file since it is missing
data = {
    "Student_ID": [101, 102, 103, 104],
    "Python": [85, 45, 70, 30],
    "Maths": [90, 50, 65, 35],
    "DBMS": [80, 55, 75, 40],
    "Attendance": [95, 75, 85, 60]
}
pd.DataFrame(data).to_csv("student_records.csv", index=False)

# Load student records
df = pd.read_csv("student_records.csv")

# Add study hours
df["Study_Hours"] = [3, 5, 1, 2]

# Calculate average marks
df["Average"] = (
    df["Python"] + df["Maths"] + df["DBMS"]
) / 3

# Create performance categories
def classify_performance(average):
    if average >= 80:
        return "Excellent"
    elif average >= 60:
        return "Good"
    elif average >= 40:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average"].apply(
    classify_performance
)

# Select input and target
X = df[["Study_Hours", "Attendance"]]
y = df["Performance"]

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Create Decision Tree model
model = DecisionTreeClassifier(random_state=42)

# Train the model
model.fit(X_train, y_train)

print("\nMachine Learning Model:")
print("Decision Tree Classifier")

print("\nModel trained successfully!")
```

**Concepts Used**
1. Decision Tree Classifier: A Machine Learning algorithm used to classify data into categories.
2. Supervised Learning: Learning from labelled examples containing inputs and expected outputs.
3. fit(): Trains the model using training data.
4. Model Training: The process of teaching a model to identify patterns.
5. Scikit-learn: Provides tools and algorithms for Machine Learning.
