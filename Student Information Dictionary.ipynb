***Student Information Dictionary***

**Objective**

To extend the project by using a dictionary to store each student's information with meaningful labels and make the student records easier to access and understand.

**Theory**

The project has already organized student information using tuples. However, accessing values using positions such as student[1] or student[2] can become difficult to understand.

A dictionary allows the project to store information using keys and values. For example, instead of remembering that position 1 represents Python marks, we can directly use the key "Python".

In this stage, each student's name and subject marks will be stored in a dictionary. The analyzer will then use these values to calculate total marks, average marks, and performance.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records using dictionaries
student_records = [
    {
        "Name": "Rahul",
        "Python": 78,
        "Maths": 82,
        "DBMS": 75
    },
    {
        "Name": "Priya",
        "Python": 88,
        "Maths": 91,
        "DBMS": 85
    },
    {
        "Name": "Aman",
        "Python": 65,
        "Maths": 70,
        "DBMS": 68
    }
]

# Process each student
print("\nStudent Performance")
print("-------------------")

for student in student_records:

    name = student["Name"]
    python_marks = student["Python"]
    maths_marks = student["Maths"]
    dbms_marks = student["DBMS"]

    # Calculate total and average
    total_marks = python_marks + maths_marks + dbms_marks
    average_marks = total_marks / 3

    # Evaluate performance
    if average_marks >= 80:
        performance = "Excellent"
    elif average_marks >= 60:
        performance = "Good"
    elif average_marks >= 40:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    print("\nStudent:", name)
    print("Python Marks:", python_marks)
    print("Maths Marks:", maths_marks)
    print("DBMS Marks:", dbms_marks)
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Performance:", performance)
```
