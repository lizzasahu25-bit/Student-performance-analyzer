***Calculating Class Statistics***

**Objective**

To extend the project by calculating the highest average, lowest average, and overall class average using the existing student records.

**Theory**

The project can now generate individual student grade reports. The next step is to understand the overall academic performance of the entire class.
In this stage, the project stores each student's average marks in a list. It then uses Python's built-in functions to calculate the highest average, lowest average, and average marks of the class.
These statistics help compare student performance and provide a simple overview of the class's academic results.

**Code**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75,
     "Attendance": 85},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85,
     "Attendance": 92},
    {"Name": "Aman", "Python": 35, "Maths": 42, "DBMS": 38,
     "Attendance": 68},
    {"Name": "Sneha", "Python": 55, "Maths": 58, "DBMS": 52,
     "Attendance": 72}
]

subjects = ["Python", "Maths", "DBMS"]

# Store student averages
student_averages = []

print("\nIndividual Student Averages")
print("----------------------------")

for student in student_records:

    total_marks = 0

    for subject in subjects:
        total_marks += student[subject]

    average_marks = total_marks / len(subjects)
    student_averages.append(average_marks)

    print(student["Name"], ":", round(average_marks, 2))

# Calculate class statistics
highest_average = max(student_averages)
lowest_average = min(student_averages)
class_average = sum(student_averages) / len(student_averages)

print("\nClass Statistics")
print("----------------")
print("Highest Average:", round(highest_average, 2))
print("Lowest Average:", round(lowest_average, 2))
print("Overall Class Average:", round(class_average, 2))
```

**Concept Used**
- Lists and append()
- max() and min()
- sum() and len()
- for loops
- Dictionaries
- Average calculation
- Rounding using round()
