***Introduction to Machine Learning***

**1. Objective**

To understand the basic concept of Machine Learning and prepare the student dataset for building a prediction model.

**2. Theory**

Machine Learning is a branch of Artificial Intelligence that allows computers to learn patterns from existing data and make predictions without being explicitly programmed for every situation.

In our project, we will use Machine Learning to predict student performance based on information such as study hours, attendance, and previous marks.

Before building a prediction model, we need to prepare our dataset.

In this project, we will:
- Load student records using Pandas.
- Select important student information as input features.
- Calculate average marks.
- Divide students into performance categories.
- Prepare the data for future Machine Learning tasks.
We will use the Scikit-learn library, which provides tools for building Machine Learning models.

**3. Code**

```python
import pandas as pd
import io

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Create sample dataset and save it as a CSV file to prevent FileNotFoundError
sample_data = """Name,Python,Maths,DBMS,Attendance
Alice,85,90,88,95
Bob,55,62,45,80
Charlie,35,40,38,60
David,72,68,75,85
"""
with open("student_records.csv", "w") as f:
    f.write(sample_data)

# Load student records
df = pd.read_csv("student_records.csv")

# Add study hours
df["Study_Hours"] = [3, 5, 1, 2]

# Calculate average marks
df["Average"] = (
    df["Python"] + df["Maths"] + df["DBMS"]
) / 3

# Classify student performance
def classify_performance(average):
    if average >= 80:
        return "Excellent"
    elif average >= 60:
        return "Good"
    elif average >= 40:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average"].apply(
    classify_performance
)

# Select input features
X = df[["Study_Hours", "Attendance"]]

# Select target output
y = df["Performance"]

# Display prepared dataset
print("\nPrepared Student Dataset:")
print(df[[
    "Name", "Study_Hours", "Attendance",
    "Average", "Performance"
]])

print("\nInput Features:")
print(X)

print("\nTarget Output:")
print(y)
```

Concepts Used
1. Machine Learning: Allows computers to learn patterns from data and make predictions.
2. Scikit-learn: A Python library used for Machine Learning.
3. Features (X): Input information used by a model to make predictions.
4. Target (y): The output that the model is trained to predict.
5. Classification: Dividing data into categories based on specific conditions.
6. apply(): Applies a function to each value in a Pandas column.
