***Analyzing CSV Data Using Pandas***

**1. Objective**

To use the Pandas library to load, organize, and analyze student records stored in a CSV file.

**2. Theory**

In the previous days, we used Python's CSV module to save and retrieve student records. However, analyzing large amounts of data using ordinary Python code can become difficult.

Pandas is a Python library that makes working with structured data easier.

In this project, we will use Pandas to:
- Load student records from a CSV file.
- Display the complete student dataset.
- View the first few records.
- Check the number of students and columns.
- Display basic information about the dataset.
Pandas stores tabular information in a structure called a DataFrame, which looks similar to a spreadsheet.

**3. Code**

```python
import pandas as pd
import os

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Create a sample student_records.csv if it does not exist
csv_filename = "student_records.csv"
if not os.path.exists(csv_filename):
    print(f"'{csv_filename}' not found. Creating a sample file...")
    sample_data = {
        "Student_ID": [101, 102, 103, 104, 105, 106],
        "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
        "Math_Score": [85, 90, 78, 92, 88, 76],
        "Science_Score": [92, 85, 80, 89, 94, 82],
        "English_Score": [88, 87, 85, 90, 91, 80]
    }
    pd.DataFrame(sample_data).to_csv(csv_filename, index=False)
    print("Sample file created successfully.")

# Load student records from CSV file
df = pd.read_csv(csv_filename)

# Display complete dataset
print("\nStudent Records:")
print(df)

# Display first five records
print("\nFirst Five Records:")
print(df.head())

# Display dataset dimensions
print("\nDataset Shape:")
print(df.shape)

# Display column information
print("\nDataset Information:")
df.info()

# Display column names
print("\nColumn Names:")
print(df.columns)
```

***Concepts Used***
1. Pandas: A Python library used for data analysis and manipulation.
2. DataFrame: A tabular structure containing rows and columns.
3. read_csv(): Loads data from a CSV file.
4. head(): Displays the first five rows by default.
5. shape: Returns the number of rows and columns.
6. info(): Displays dataset structure and data types.
7. columns: Displays the names of all columns.
