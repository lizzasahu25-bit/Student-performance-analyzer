***Statistical Analysis Using Pandas***

**1. Objective**

To calculate important statistical values from student records using Pandas and identify the overall academic performance of the class.

**2. Theory**

After loading student records into a Pandas DataFrame, we can perform different calculations to understand the dataset.

Statistical analysis helps us identify the average marks, highest marks, lowest marks, and overall performance in each subject.

In this project, we will use Pandas to:
- Calculate average marks in each subject.
- Find the highest and lowest marks.
- Calculate average attendance.
- Generate a statistical summary of the dataset.
These calculations will help us understand which subjects have better performance and where students need improvement.

**3. Code**

```python
import pandas as pd

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Load student records
df = pd.read_csv("student_records.csv")

# Calculate subject-wise averages
print("\nSubject-Wise Average Marks:")

print(df[["Python", "Maths", "DBMS"]].mean())

# Find highest marks
print("\nHighest Marks in Each Subject:")

print(df[["Python", "Maths", "DBMS"]].max())

# Find lowest marks
print("\nLowest Marks in Each Subject:")

print(df[["Python", "Maths", "DBMS"]].min())

# Calculate average attendance
average_attendance = df["Attendance"].mean()

print("\nAverage Class Attendance:",
      round(average_attendance, 2), "%")

# Generate statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

Concepts Used
1. mean(): Calculates the average value.
2. max(): Finds the highest value.
3. min(): Finds the lowest value.
4. describe(): Generates a statistical summary of numerical columns.
5. Column Selection: Selects particular subject columns for analysis.
