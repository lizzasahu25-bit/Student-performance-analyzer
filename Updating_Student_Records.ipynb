***Updating Student Records***

**1. Objective**

To allow users to search for an existing student and update their marks or attendance in the project database.

**2. Theory**

Sometimes a student's marks or attendance may need correction after a record has been added.

In this project, we will create an update feature that:
- Searches for a student using their name.
- Displays the student's existing information.
- Allows the user to enter updated marks and attendance.
- Replaces the old values with the new values.
We will use a for loop to search records, a while loop for input validation, and dictionary keys to update student information.

**3. Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Existing student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82,
     "DBMS": 75, "Attendance": 85},

    {"Name": "Priya", "Python": 88, "Maths": 91,
     "DBMS": 85, "Attendance": 92},

    {"Name": "Aman", "Python": 35, "Maths": 42,
     "DBMS": 38, "Attendance": 68},

    {"Name": "Sneha", "Python": 55, "Maths": 58,
     "DBMS": 52, "Attendance": 72}
]

# Function for input validation
def get_valid_input(message):
    while True:
        value = int(input(message))

        if 0 <= value <= 100:
            return value
        else:
            print("Enter a value between 0 and 100.")

# Search for student
search_name = input("\nEnter student name to update: ").strip()

student_found = False

for student in student_records:

    if student["Name"].lower() == search_name.lower():

        student_found = True

        print("\nExisting Student Details")
        print("Name:", student["Name"])
        print("Python:", student["Python"])
        print("Maths:", student["Maths"])
        print("DBMS:", student["DBMS"])
        print("Attendance:", student["Attendance"], "%")

        print("\nEnter Updated Details")

        # Update student information
        student["Python"] = get_valid_input("Enter new Python marks: ")
        student["Maths"] = get_valid_input("Enter new Maths marks: ")
        student["DBMS"] = get_valid_input("Enter new DBMS marks: ")
        student["Attendance"] = get_valid_input(
            "Enter new attendance percentage: "
        )

        print("\nStudent record updated successfully!")

        print("\nUpdated Student Details")
        print("Name:", student["Name"])
        print("Python:", student["Python"])
        print("Maths:", student["Maths"])
        print("DBMS:", student["DBMS"])
        print("Attendance:", student["Attendance"], "%")

        break

if not student_found:
    print("\nStudent record not found.")
```

**Concepts Used**
1. For Loop: Searches for a particular student in the records.
2. Dictionary Updating: Changes existing values using dictionary keys.
3. Functions: Reuses the input validation function.
4. Break Statement: Stops searching once the student is found.
5. Case-Insensitive Search: Allows names to be searched regardless of uppercase or lowercase letters.
