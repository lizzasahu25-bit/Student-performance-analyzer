***Finding the Lowest-Scoring Subject***

**Objective**

To extend the project by identifying the subject with the lowest class average and highlighting where students may need more practice.

**Theory**

The project can now calculate the average marks for each subject across the class. The next step is to identify the subject with the lowest average.
In this stage, the project uses the min() function to find the subject with the lowest class average from the existing dictionary of subject averages. This helps identify a subject that may require additional revision or teaching support.
Together with Day 13's highest-average analysis, this feature gives a clearer picture of the class's academic strengths and weaknesses.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85},
    {"Name": "Aman", "Python": 35, "Maths": 42, "DBMS": 38},
    {"Name": "Sneha", "Python": 55, "Maths": 58, "DBMS": 52}
]

# Subject names
subjects = ["Python", "Maths", "DBMS"]

# Calculate subject-wise averages
subject_averages = {}

for subject in subjects:
    total = 0

    for student in student_records:
        total += student[subject]

    average = total / len(student_records)
    subject_averages[subject] = average

# Display subject averages
print("\nSubject-Wise Class Average")
print("--------------------------")

for subject, average in subject_averages.items():
    print(subject + ":", round(average, 2))

# Find highest and lowest averages
best_subject = max(subject_averages, key=subject_averages.get)
weakest_subject = min(subject_averages, key=subject_averages.get)

print("\nClass Performance Analysis")
print("--------------------------")
print("Highest-Average Subject:", best_subject)
print("Average Marks:", round(subject_averages[best_subject], 2))

print("\nSubject Needing More Attention")
print("------------------------------")
print("Subject:", weakest_subject)
print("Average Marks:", round(subject_averages[weakest_subject], 2))
```

Concept Used
- min() function
- max() function
- Dictionaries
- Nested for loops
- len() function
- Subject-wise average calculation
