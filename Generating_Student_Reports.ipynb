***Generating Student Reports***

**1. Objective**

To generate a complete academic report for each student, including total marks, average marks, grade, pass/fail status, and attendance.

**2. Theory**
After adding, updating, and deleting student records, the project needs a feature to display their academic performance in an organized format.

A student report helps us understand individual performance without manually checking each subject's marks.

In this project, we will:
- Calculate total and average marks.
- Assign grades based on average marks.
- Check whether the student has passed or failed.
- Display attendance percentage.
- Generate reports for all available students.
We will use functions to organize the report-generation process and avoid repeating the same calculations.

**3. Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82,
     "DBMS": 75, "Attendance": 85},

    {"Name": "Priya", "Python": 88, "Maths": 91,
     "DBMS": 85, "Attendance": 92},

    {"Name": "Aman", "Python": 35, "Maths": 42,
     "DBMS": 38, "Attendance": 68},

    {"Name": "Sneha", "Python": 55, "Maths": 58,
     "DBMS": 52, "Attendance": 72}
]

# Function to generate student report
def generate_report(student):

    total_marks = (
        student["Python"] +
        student["Maths"] +
        student["DBMS"]
    )

    average_marks = total_marks / 3

    # Grade calculation
    if average_marks >= 80:
        grade = "A"
    elif average_marks >= 70:
        grade = "B"
    elif average_marks >= 60:
        grade = "C"
    elif average_marks >= 40:
        grade = "D"
    else:
        grade = "F"

    # Pass/Fail calculation
    if (student["Python"] >= 40 and
        student["Maths"] >= 40 and
        student["DBMS"] >= 40):
        result = "Pass"
    else:
        result = "Fail"

    # Display report
    print("\nStudent Name:", student["Name"])
    print("------------------------------")
    print("Python Marks:", student["Python"])
    print("Maths Marks:", student["Maths"])
    print("DBMS Marks:", student["DBMS"])
    print("Total Marks:", total_marks, "/ 300")
    print("Average Marks:", round(average_marks, 2))
    print("Grade:", grade)
    print("Result:", result)
    print("Attendance:", student["Attendance"], "%")


# Generate reports for all students
for student in student_records:
    generate_report(student)
```

Concepts Used
1. Functions: Organize report-generation tasks into a reusable block.
2. Parameters: Pass individual student records to the function.
3. Return-free Function: Displays the report directly.
4. Conditional Statements: Determine grades and pass/fail results.
5. Round() Function: Displays average marks up to two decimal places.
