***Visualizing Study Hours and Marks***

**1. Objective**

To analyze and visualize the relationship between students' study hours and their average marks using a scatter plot.

**2. Theory**

Until now, we have analyzed student performance using marks, attendance, and subject-wise averages.

However, study time is also an important factor that may influence academic performance.

A Scatter Plot is a type of graph that uses individual points to represent the relationship between two numerical variables.

In this project, we will:
- Store the study hours of each student.
- Calculate their average marks.
- Create a scatter plot comparing study hours and average marks.
- Observe whether students who study longer tend to achieve higher marks.
This analysis will help us understand the possible relationship between study habits and academic performance.

**3. Code**

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Create sample student_records.csv if it does not exist
if not os.path.exists("student_records.csv"):
    sample_data = {
        "Name": ["Alice", "Bob", "Charlie", "David"],
        "Python": [85, 92, 70, 78],
        "Maths": [90, 88, 65, 80],
        "DBMS": [80, 95, 60, 85]
    }
    pd.DataFrame(sample_data).to_csv("student_records.csv", index=False)

# Load student records
df = pd.read_csv("student_records.csv")

# Add study hours for each student
df["Study_Hours"] = [3, 5, 1, 2]

# Calculate average marks
df["Average"] = (
    df["Python"] + df["Maths"] + df["DBMS"]
) / 3

# Display student study hours and averages
print("\nStudy Hours and Average Marks:")
print(df[["Name", "Study_Hours", "Average"]])

# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Average"]
)

# Add chart details
plt.xlabel("Daily Study Hours")
plt.ylabel("Average Marks")
plt.title("Relationship Between Study Hours and Marks")

# Display chart
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Concepts Used**

1. Scatter Plot: Displays the relationship between two numerical variables.
2. scatter(): Creates a scatter plot using Matplotlib.
3. DataFrame Column: Stores study hours and average marks.
4. xlabel() and ylabel(): Add labels to the graph axes.
5. grid(): Adds grid lines to improve graph readability.
6. show(): Displays the generated graph.
