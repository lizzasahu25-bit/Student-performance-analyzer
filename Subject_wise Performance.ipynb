***Subject-Wise Performance***

**Objective**

To extend the project by using a set to identify the subjects included in the student's performance record and display the available subjects without repetition.

**Theory**

The project currently stores each student's marks for Python, Maths, and DBMS. As the project grows, it may contain many subject names, and duplicate subject names should not be stored unnecessarily.

A set is useful when we need to store unique values. In this stage, the project uses a set to maintain the subjects being analyzed. The student records are then processed using the existing loop and conditional logic.

This prepares the project for handling a larger number of subjects while keeping subject names unique.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Subjects analyzed in the project
subjects = {"Python", "Maths", "DBMS"}

print("\nSubjects Analyzed")
print("-----------------")

for subject in subjects:
    print(subject)

# Student records
student_records = [
    ("Rahul", 78, 82, 75),
    ("Priya", 88, 91, 85),
    ("Aman", 65, 70, 68)
]

# Process each student
print("\nStudent Performance")
print("-------------------")

for student in student_records:

    name = student[0]
    python_marks = student[1]
    maths_marks = student[2]
    dbms_marks = student[3]

    # Calculate total and average
    total_marks = python_marks + maths_marks + dbms_marks
    average_marks = total_marks / 3

    # Evaluate performance
    if average_marks >= 80:
        performance = "Excellent"
    elif average_marks >= 60:
        performance = "Good"
    elif average_marks >= 40:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    print("\nStudent:", name)
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Performance:", performance)
```
