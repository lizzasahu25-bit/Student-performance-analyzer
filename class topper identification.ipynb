***Finding the Class Topper***

**Objective**

To extend the project by identifying the student with the highest average marks and displaying the class topper.

**Theory**

The project can now analyze individual students and summarize the class performance. The next step is to identify the student who achieved the highest average marks.
In this stage, the project uses a variable called topper_average to store the highest average found so far and topper_name to store that student's name. During each loop iteration, the student's average is compared with the highest average. If it is higher, the topper information is updated.
This feature helps the analyzer identify the top-performing student in the class.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85},
    {"Name": "Aman", "Python": 65, "Maths": 70, "DBMS": 68}
]

# Initialize topper details
topper_name = ""
topper_average = -1

print("\nStudent Performance")
print("-------------------")

# Analyze each student
for student in student_records:

    name = student["Name"]

    total_marks = (
        student["Python"]
        + student["Maths"]
        + student["DBMS"]
    )

    average_marks = total_marks / 3

    if average_marks >= 80:
        performance = "Excellent"
    elif average_marks >= 60:
        performance = "Good"
    elif average_marks >= 40:
        performance = "Average"
    else:
        performance = "Needs Improvement"

    print("\nStudent:", name)
    print("Total Marks:", total_marks)
    print("Average Marks:", round(average_marks, 2))
    print("Performance:", performance)

    # Find the class topper
    if average_marks > topper_average:
        topper_average = average_marks
        topper_name = name

# Display topper
print("\nClass Topper")
print("------------")
print("Name:", topper_name)
print("Average Marks:", round(topper_average, 2))
```

**Concept Used**
- for loop
- Dictionaries
- Comparison operator (>)
- Conditional statements
- Updating variables
- round() function
