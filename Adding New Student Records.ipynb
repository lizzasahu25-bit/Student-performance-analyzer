***Adding New Student Records***

**Objective**

To extend the project by allowing users to enter a new student's details and add the record to the existing student database.

**Theory**

The project can already analyze student records and search for individual students. However, the records are currently written directly into the code, so adding a new student requires editing the program.
In this stage, the project uses the input() function to collect a student's name, subject marks, and attendance. The int() function converts the entered marks and attendance into numbers, while the append() method adds the new student's dictionary to the existing list.
This makes the project more interactive by allowing users to add student records during program execution.

**Code**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Existing student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82,
     "DBMS": 75, "Attendance": 85},
    {"Name": "Priya", "Python": 88, "Maths": 91,
     "DBMS": 85, "Attendance": 92},
    {"Name": "Aman", "Python": 35, "Maths": 42,
     "DBMS": 38, "Attendance": 68},
    {"Name": "Sneha", "Python": 55, "Maths": 58,
     "DBMS": 52, "Attendance": 72}
]

# Enter new student details
print("\nAdd New Student")
print("----------------")

name = input("Enter student name: ").strip()

python_marks = int(input("Enter Python marks: "))
maths_marks = int(input("Enter Maths marks: "))
dbms_marks = int(input("Enter DBMS marks: "))
attendance = int(input("Enter attendance percentage: "))

# Create a new student record
new_student = {
    "Name": name,
    "Python": python_marks,
    "Maths": maths_marks,
    "DBMS": dbms_marks,
    "Attendance": attendance
}

# Add the record
student_records.append(new_student)

print("\nStudent added successfully!")
print("Name:", new_student["Name"])
print("Python Marks:", new_student["Python"])
print("Maths Marks:", new_student["Maths"])
print("DBMS Marks:", new_student["DBMS"])
print("Attendance:", new_student["Attendance"], "%")

print("\nTotal Student Records:", len(student_records))
```

**Concept Used**
- input() function
- int() type conversion
- Dictionaries
- Lists and append()
- len() function
- User input and record creation
