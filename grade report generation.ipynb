***Generating Student Grade Reports***

**Objective**

To extend the project by assigning a grade to each student based on their average marks and displaying a complete academic report.

**Theory**

The project can now calculate average marks, check attendance, and evaluate overall student status. The next step is to make the academic results easier to understand by assigning grades.
In this stage, the project uses conditional statements to assign grades based on average marks. For example, an average of 80 or above receives Grade A, while lower averages receive other grades according to predefined ranges.
The grade report will help students and teachers understand academic performance quickly.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {
        "Name": "Rahul",
        "Python": 78,
        "Maths": 82,
        "DBMS": 75,
        "Attendance": 85
    },
    {
        "Name": "Priya",
        "Python": 88,
        "Maths": 91,
        "DBMS": 85,
        "Attendance": 92
    },
    {
        "Name": "Aman",
        "Python": 35,
        "Maths": 42,
        "DBMS": 38,
        "Attendance": 68
    },
    {
        "Name": "Sneha",
        "Python": 55,
        "Maths": 58,
        "DBMS": 52,
        "Attendance": 72
    }
]

subjects = ["Python", "Maths", "DBMS"]

print("\nStudent Grade Reports")
print("---------------------")

for student in student_records:

    name = student["Name"]

    # Calculate total and average
    total_marks = 0

    for subject in subjects:
        total_marks += student[subject]

    average_marks = total_marks / len(subjects)

    # Assign grade
    if average_marks >= 80:
        grade = "A"
    elif average_marks >= 70:
        grade = "B"
    elif average_marks >= 60:
        grade = "C"
    elif average_marks >= 40:
        grade = "D"
    else:
        grade = "F"

    # Check pass/fail status
    passed_all_subjects = True

    for subject in subjects:
        if student[subject] < 40:
            passed_all_subjects = False

    if passed_all_subjects:
        result = "Pass"
    else:
        result = "Fail"

    print("\nStudent:", name)
    print("Total Marks:", total_marks)
    print("Average Marks:", round(average_marks, 2))
    print("Grade:", grade)
    print("Result:", result)
    print("Attendance:", student["Attendance"], "%")
```

**Concept Used**

- Conditional statements (if, elif, else)
- Dictionaries and lists
- for loops
- Grade classification
- Arithmetic operators
- Pass/fail evaluation
