**Student Performance Evaluation**

***Objective***

To extend the project by evaluating the student's average marks and assigning a performance category using conditional statements.

***Theory***

After calculating the student's average marks, the project needs to interpret the result. A numerical average alone does not tell us whether the student's performance is good, average, or needs improvement.

Conditional statements allow the program to make decisions based on a condition. In this project, the student's average marks are compared with predefined ranges using if, elif, and else.

The program will classify the student's performance into categories such as Excellent, Good, Average, or Needs Improvement. This will form the basic decision-making part of the analyzer.

***Code:***

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student information
student_name = "Rahul"
attendance = 85
study_hours = 4.5

# Subject marks
python_marks = 78
maths_marks = 82
dbms_marks = 75

# Calculate total and average
total_marks = python_marks + maths_marks + dbms_marks
average_marks = total_marks / 3

# Evaluate performance
if average_marks >= 80:
    performance = "Excellent"
elif average_marks >= 60:
    performance = "Good"
elif average_marks >= 40:
    performance = "Average"
else:
    performance = "Needs Improvement"

print("\nStudent Information")
print("-------------------")
print("Name:", student_name)
print("Attendance:", attendance, "%")
print("Study Hours:", study_hours)

print("\nAcademic Performance")
print("--------------------")
print("Python Marks:", python_marks)
print("Maths Marks:", maths_marks)
print("DBMS Marks:", dbms_marks)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Performance:", performance)
```

***Concept Used***
if → checks the first condition.

elif → checks additional conditions if the previous condition is false.

else → handles all remaining cases.

Comparison operators (>=) → compare the average marks with performance thresholds.

Conditional decision-making → converts the calculated average into a meaningful performance category.
