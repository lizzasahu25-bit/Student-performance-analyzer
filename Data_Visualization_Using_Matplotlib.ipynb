***Data Visualization Using Matplotlib***

**1. Objective**

To create visual representations of student marks using Matplotlib and understand the performance differences between students.

**2. Theory**

After analyzing student records using Pandas, the next step is to represent the information visually.

Data visualization helps us understand student performance more easily through graphs and charts.

In this project, we will use Matplotlib, a Python library used for creating different types of graphs.

We will create a bar chart to compare the marks obtained by different students in Python, Maths, and DBMS.

The chart will help us identify which students perform better in each subject.

**3. Code**

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Create a sample student_records.csv if it does not exist
if not os.path.exists("student_records.csv"):
    sample_data = {
        "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
        "Python": [85, 72, 90, 68, 95],
        "Maths": [78, 88, 81, 74, 91],
        "DBMS": [92, 80, 85, 70, 88]
    }
    pd.DataFrame(sample_data).to_csv("student_records.csv", index=False)
    print("Created a sample 'student_records.csv' file.")

# Load student records
df = pd.read_csv("student_records.csv")

# Select student names and subject marks
students = df["Name"]

python_marks = df["Python"]
maths_marks = df["Maths"]
dbms_marks = df["DBMS"]

# Set bar chart positions
x = range(len(students))
width = 0.25

# Create bar chart
plt.figure(figsize=(9, 5))

plt.bar(x, python_marks, width, label="Python")
plt.bar([i + width for i in x],
        maths_marks, width, label="Maths")
plt.bar([i + 2 * width for i in x],
        dbms_marks, width, label="DBMS")

# Add chart details
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Subject-Wise Student Performance")

plt.xticks(
    [i + width for i in x],
    students
)

plt.legend()
plt.tight_layout()

# Display chart
plt.show()
```

**Concepts Used**
1. Matplotlib: A Python library used for data visualization.
2. pyplot: A Matplotlib module used to create graphs.
3. bar(): Creates a bar chart.
4. xlabel() and ylabel(): Add labels to the graph axes.
5. title(): Adds a title to the chart.
6. legend(): Identifies different categories in the graph.
7. show(): Displays the generated chart.
