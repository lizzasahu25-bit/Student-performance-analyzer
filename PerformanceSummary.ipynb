***Performance Summary***

**Objective**

To extend the project by using the existing student records to calculate and display a summary of the class performance.

**Theory**

The project can now store and analyze information for multiple students. The next step is to summarize the results instead of looking at each student separately.

In this stage, the project will use the results calculated for each student to count how many students achieved each performance category. This will help the analyzer provide a simple overview of the class, such as how many students performed excellently, well, or need improvement.

The for loop is used to check every student record, while the existing conditional statements classify their performance.

**Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {
        "Name": "Rahul",
        "Python": 78,
        "Maths": 82,
        "DBMS": 75
    },
    {
        "Name": "Priya",
        "Python": 88,
        "Maths": 91,
        "DBMS": 85
    },
    {
        "Name": "Aman",
        "Python": 65,
        "Maths": 70,
        "DBMS": 68
    }
]

# Counters for performance categories
excellent_count = 0
good_count = 0
average_count = 0
needs_improvement_count = 0

print("\nStudent Performance")
print("-------------------")

# Analyze each student
for student in student_records:

    name = student["Name"]

    total_marks = (
        student["Python"]
        + student["Maths"]
        + student["DBMS"]
    )

    average_marks = total_marks / 3

    # Classify performance
    if average_marks >= 80:
        performance = "Excellent"
        excellent_count += 1

    elif average_marks >= 60:
        performance = "Good"
        good_count += 1

    elif average_marks >= 40:
        performance = "Average"
        average_count += 1

    else:
        performance = "Needs Improvement"
        needs_improvement_count += 1

    print("\nStudent:", name)
    print("Total Marks:", total_marks)
    print("Average Marks:", average_marks)
    print("Performance:", performance)

# Display class summary
print("\nClass Performance Summary")
print("-------------------------")
print("Excellent:", excellent_count)
print("Good:", good_count)
print("Average:", average_count)
print("Needs Improvement:", needs_improvement_count)
```
