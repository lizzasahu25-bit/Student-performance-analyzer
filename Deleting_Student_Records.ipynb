**Deleting Student Records**

**1. Objective**

To allow users to search for a student and delete their record from the project database.

**2. Theory**

Sometimes a student record may be entered incorrectly or may no longer be required. In such cases, the project should provide an option to remove that record.

In this project, we will create a delete feature that:
- Takes the student's name as input.
- Searches for the student in the existing records.
- Deletes the matching record.
- Displays a confirmation message.
We will use the remove() method to delete a student's dictionary from the list of records.

**3. Code:**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Existing student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82,
     "DBMS": 75, "Attendance": 85},

    {"Name": "Priya", "Python": 88, "Maths": 91,
     "DBMS": 85, "Attendance": 92},

    {"Name": "Aman", "Python": 35, "Maths": 42,
     "DBMS": 38, "Attendance": 68},

    {"Name": "Sneha", "Python": 55, "Maths": 58,
     "DBMS": 52, "Attendance": 72}
]

# Display existing student records
print("\nExisting Students:")

for student in student_records:
    print(student["Name"])

# Search for a student to delete
search_name = input("\nEnter student name to delete: ").strip()

student_found = False

for student in student_records:

    if student["Name"].lower() == search_name.lower():

        student_found = True

        print("\nStudent Found:", student["Name"])

        # Delete student record
        student_records.remove(student)

        print("Student record deleted successfully!")

        break

if not student_found:
    print("\nStudent record not found.")

# Display remaining records
print("\nRemaining Student Records:")

for student in student_records:
    print(student["Name"])

print("\nTotal Students:", len(student_records))
```

**Concepts Used**
1. List: Stores multiple student records.
2. For Loop: Searches for a particular student.
3. Remove() Method: Deletes the selected student record from the list.
4. Conditional Statements: Checks whether the student exists.
5. Break Statement: Stops the search after deleting the record.
