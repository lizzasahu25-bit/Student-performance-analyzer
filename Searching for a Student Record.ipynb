***Searching for a Student Record***

**Objective**

To extend the project by allowing users to search for a particular student by name and view their academic performance and attendance details.
**Theory**

The project can now analyze multiple students and rank them according to their average marks. However, users may want to view the details of one specific student without reading the entire report.
In this stage, the project accepts a student's name as input and searches for a matching record. The input() function collects the name, and a for loop checks each student's record. If a match is found, the program displays the student's marks, average, and attendance.
This feature makes the analyzer more interactive and convenient to use.

**Code**

```python
print("Student Performance Analyzer & Predictor")
print("----------------------------------------")

# Student records
student_records = [
    {"Name": "Rahul", "Python": 78, "Maths": 82, "DBMS": 75,
     "Attendance": 85},
    {"Name": "Priya", "Python": 88, "Maths": 91, "DBMS": 85,
     "Attendance": 92},
    {"Name": "Aman", "Python": 35, "Maths": 42, "DBMS": 38,
     "Attendance": 68},
    {"Name": "Sneha", "Python": 55, "Maths": 58, "DBMS": 52,
     "Attendance": 72}
]

subjects = ["Python", "Maths", "DBMS"]

# Search for a student
search_name = input("\nEnter student name to search: ").strip()

student_found = False

for student in student_records:

    if student["Name"].lower() == search_name.lower():

        student_found = True

        total_marks = 0

        for subject in subjects:
            total_marks += student[subject]

        average_marks = total_marks / len(subjects)

        print("\nStudent Record Found")
        print("--------------------")
        print("Name:", student["Name"])
        print("Python Marks:", student["Python"])
        print("Maths Marks:", student["Maths"])
        print("DBMS Marks:", student["DBMS"])
        print("Total Marks:", total_marks)
        print("Average Marks:", round(average_marks, 2))
        print("Attendance:", student["Attendance"], "%")

        break

if not student_found:
    print("\nStudent not found. Please check the name.")
```

**Concept Used**
- input() function
- String methods: strip() and lower()
- for loop
- Conditional statements
- Boolean variable
- break statement
- Dictionary access
